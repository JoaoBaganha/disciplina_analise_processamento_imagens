```python
ws_bgr = cv2.imread('ws.jpg')
ws_rgb = cv2.cvtColor(ws_bgr, cv2.COLOR_BGR2RGB)
ws_hsv = cv2.cvtColor(ws_bgr, cv2.COLOR_BGR2HSV)
ws_gray = cv2.cvtColor(ws_bgr, cv2.COLOR_BGR2GRAY)
ws_hls = cv2.cvtColor(ws_bgr, cv2.COLOR_BGR2HLS)
```

```python
ws_hls
```

```python
plt.figure(figsize=(8,5))
plt.title('Safadão BGR')
plt.imshow(ws_bgr)
```

```python
plt.subplots(figsize=(8,5))
plt.title('Safadão RGB')
plt.imshow(cv2.cvtColor(ws_bgr, cv2.COLOR_BGR2RGB))
```

```python
fig, ax = plt.subplots(1, 5, figsize=(15,15))

ax[0].imshow(ws_bgr)
ax[0].set_title('Safadão BGR')

ax[1].imshow(ws_rgb)
ax[1].set_title('Safadão RGB')

ax[2].imshow(ws_hsv)
ax[2].set_title('Safadão HSV')

ax[3].imshow(ws_gray, cmap='gray')
ax[3].set_title('Safadão GRAY')

ax[4].imshow(ws_hls)
ax[4].set_title('Safadão HLS')

plt.savefig('conversao_cores_safadaaaaaao.png')
plt.show()
```

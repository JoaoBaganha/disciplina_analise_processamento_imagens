```python
from google.colab import files
import cv2
from google.colab.patches import cv2_imshow
import numpy as np
import matplotlib.pyplot as plt
```

```python
ligaDaJusticaBgr = cv2.imread('ligaDaJustica.jpg')
```

```python
ligaDaJusticaGray = cv2.cvtColor(ligaDaJusticaBgr, cv2.COLOR_BGR2GRAY)
```

```python
cv2_imshow(ligaDaJusticaGray)
```

```python
def calcularHist(img):
  imgVetor = img.ravel()
  histVetor = np.zeros(256, dtype = np.int64)
  for px in imgVetor:
    histVetor[px] += 1

  return histVetor


```

```python
calcularHist(ligaDaJusticaGray)
```

```python
histImg = calcularHist(ligaDaJusticaGray)
plt.bar(range(256), histImg)
plt.show
```

```python
def limiarImg (img, limiarInferior, limiarSuperior):
  imgVetor = img.ravel()
  histVetor = np.zeros(256, dtype = np.int64)
  for px in imgVetor:
    if limiarInferior <= px <= limiarSuperior:
      histVetor[px] +=1

  return histVetor
```

```python
imgLimiar1 = limiarImg(ligaDaJusticaGray, 0, 150)
```

```python
cv2_imshow(imgLimiar1)
```

```python
foto_exposicao_boa = cv2.imread('foto_exposicao_boa.jpeg')
foto_exposicao_clara = cv2.imread('foto_exposicao_clara.jpeg')
foto_exposicao_escura = cv2.imread('foto_exposicao_escura.jpeg')
```

```python
foto_exposicao_boa_gray = cv2.cvtColor(foto_exposicao_boa, cv2.COLOR_BGR2GRAY)
foto_exposicao_clara_gray = cv2.cvtColor(foto_exposicao_clara, cv2.COLOR_BGR2GRAY)
foto_exposicao_escura_gray = cv2.cvtColor(foto_exposicao_escura, cv2.COLOR_BGR2GRAY)
```

```python
cv2_imshow(foto_exposicao_boa_gray)
```

```python
cv2_imshow(foto_exposicao_clara_gray)
```

```python
cv2_imshow(foto_exposicao_escura_gray)
```

```python
def calcularHist(img):
  imgVetor = img.ravel()
  histVetor = np.zeros(256, dtype = np.int64)
  for px in imgVetor:
    histVetor[px] += 1

  return histVetor
```

```python
calcularHist(foto_exposicao_boa_gray)
```

```python
calcularHist(foto_exposicao_clara_gray)
```

```python
calcularHist(foto_exposicao_escura_gray)
```

```python
hist_foto_boa = calcularHist(foto_exposicao_boa)
plt.bar(range(256), hist_foto_boa)
plt.title('Histograma Foto Boa')
plt.show
```

```python
hist_foto_clara = calcularHist(foto_exposicao_clara)
plt.bar(range(256), hist_foto_clara)
plt.title('Histograma Foto Clara')
plt.show
```

```python
hist_foto_escura = calcularHist(foto_exposicao_escura)
plt.bar(range(256), hist_foto_escura)
plt.title('Histograma Foto Escura')
plt.show
```
